```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_field(filename, title, output_png):
    data = np.loadtxt(filename)
    x = data[:, 0]
    y = data[:, 1]
    u = data[:, 2]

    Nx = len(np.unique(x))
    Ny = len(np.unique(y))

    X = x.reshape((Nx, Ny))
    Y = y.reshape((Nx, Ny))
    U = u.reshape((Nx, Ny))

    plt.figure(figsize=(8, 6))
    contour = plt.contourf(X, Y, U, 100, cmap='hot')
    plt.colorbar(contour, label='Temp')
    plt.title(title)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.savefig(output_png, dpi=300)
    plt.close()
```

```python
plot_field("temperature_t0.txt", "u(t), t = 0", "temperature_t0.png")
plot_field("temperature_t_last.txt", "u(t), t = T", "temperature_t_last.png")
```
